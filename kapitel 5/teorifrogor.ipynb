# Kapitel 5 - Dimensionsreducering


1. Vad menas med curse of dimensionality

Det är de problem som uppstår när ett dataset har för många variabler (dimensioner).

Varför det blir ett problem:

Gles data (Sparsity): När dimensionerna ökar blir utrymmet så enormt att datapunkterna hamnar långt ifrån varandra. Det gör att algoritmer som bygger på avstånd (t.ex. k-NN) fungerar mycket sämre.

Ökad risk för overfitting: Modellen får så många frihetsgrader att den lätt lär sig slumpmässigt "brus" i stället för verkliga mönster.

Tyngre beräkningar: Det krävs exponentiellt mer data och datorkraft för att hantera modellen.

2. Vad är dimensionsreducering och varför görs det?

Processen att minska antalet variabler (dimensioner) i ett dataset samtidigt som man behåller så mycket som möjligt av den viktiga informationen (variansen).

3. Förklara översiktligt hur PCA fungerar. Använd figur 5.4 på sidan 224 i
din förklaring.

PCA (Principal Component Analysis): Den absolut vanligaste metoden för dimensionsreducering. Den komprimerar ner data genom att hitta nya, sammanhängande axlar som bäst fångar upp variationen.

Hur det fungerar (utifrån hur en typisk PCA-projektionsfigur ser ut):

Hitta maximal spridning (varians): Algoritmen ritar en rät linje (första huvudkomponenten, eller PC1) genom datan på det ledd där datapunkterna är som mest utspridda. Denna riktning innehåller den viktigaste informationen.

Vinkelräta axlar: Den letar sedan upp nästa axel (PC2) som är helt vinkelrät (ortogonal) mot den första och fångar upp den näst största återstående spridningen.

Projektion: Alla datapunkter "dras" (projiceras) vinkelrätt ner på denna nya axel.

Genom att göra så här kan du t.ex. gå från 2 dimensioner ner till 1 dimension (genom att bara behålla PC1) och ändå behålla merparten av informationen och mönstren i datan.
