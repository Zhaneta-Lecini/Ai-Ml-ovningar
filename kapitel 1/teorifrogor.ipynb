3. Förklara följande:**

**a) Vad är syftet med att dela upp data i träningsdata, valideringsdata och testdata?**
* Syftet är att bygga en modell som fungerar bra på helt ny data, och förhindra att den bara memorerar den data den tränas på (överanpassning). Genom uppdelningen får vi en ärlig utvärdering:

*   **Träningsdata:** Används för att själva inlärningen ska ske (modellen läser kurslitteraturen).
*   **Valideringsdata:** Används för att finjustera modellens inställningar och jämföra olika modeller mot varandra under utvecklingens gång (som ett övningsprov).
*   **Testdata:** Läggs undan helt och hållet från början. Den används *endast en enda gång* allra sist i projektet för att få ett opartiskt betyg på hur väl den färdiga modellen faktiskt fungerar på osedd data (som det slutgiltiga Nationella Provet).


**b) Vad är k-delad korsvalidering för något?**
* K-delad korsvalidering (k-fold cross-validation) är en METOD för att utvärdera en modell på ett säkrare och mer robust sätt än att bara göra en enda uppdelning av tränings- och valideringsdata.

Så här fungerar det:
1. Man delar in sin data i ett antal lika stora delar, så kallade "veck" (folds). Bokstaven *k* står för antalet delar (t.ex. k=5 betyder 5 lika stora delar).
2. Modellen tränas sedan *k* antal gånger. Varje gång används en ny, unik del som valideringsdata, medan de övriga delarna slås ihop och används som träningsdata.
3. Till slut räknar man ut ett genomsnitt av resultaten från alla *k* omgångar.

Fördelen med detta är att varje enskild datapunkt får chansen att vara valideringsdata exakt en gång. Det minskar risken att vi råkar få ett missvisande resultat bara för att vi hade "tur" eller "otur" med hur datan slumpades vid en vanlig uppdelning.

**d) Vad är en hyperparameter för något? Vad är en parameter för något?**
* Den stora skillnaden ligger i vem (eller vad) som bestämmer värdet:

*   **Parameter ("Vad"):** Detta är de inre värden och matematiska vikter som modellen *själv* räknar ut och lär sig utifrån träningsdatan. Vi kan inte ställa in dessa i förväg. Ett exempel är själva lutningen på linjen som en linjär regression räknar ut.
*   **Hyperparameter ("Hur"):** Detta är de övergripande inställningarna (eller "rattarna") som *vi programmerare* ställer in *innan* själva träningsprocessen börjar. De styr hur modellen ska lära sig. Ett exempel är hur djupt ett beslutsträd får lov att växa (`max_depth`), eller hur hårt straffet ska vara vid regularisering.

**e) Vad är grid search och hur hänger namnet “grid” och “search” ihop med processen som genomförs? [...] hyperparametern refit har standardvärdet True, vad innebär det?**
 
* **Grid Search:** Det är en METOD för att automatiskt HITTA den BÄSTA KOMBINATIONEN av HYPERRARAMETRAR för en modell.
* **Namnet:** Ordet "Grid" (rutnät) kommer från att man ställer upp en tabell/ett rutnät med alla de olika värden på inställningar man vill testa. Ordet "Search" (sök) kommer från att algoritmen systematiskt söker igenom detta nät genom att testa varenda möjlig kombination.
* **`refit=True`:** När sökandet är klart och algoritmen har korat den vinnande kombinationen av inställningar (den som gav minst fel), innebär `refit=True` att den automatiskt tränar om (refit) modellen en sista gång. Den gör detta med vinnarinställningarna på *hela* det tillgängliga datasetet (träning + validering). Resultatet är att modellen är maximalt tränad och direkt redo att användas för prediktioner med `.predict()`.

**f) Vad är kategorisk data och hur hanteras det? I ditt svar, använd begreppen nominal data, ordinal data, one-hot-encoding, dummy-variable-encoding och ordinal encoding.**
* Kategorisk data är information som beskriver grupper eller klasser istället för siffror (till exempel färger, bilmärken eller storlekar). Eftersom maskininlärningsmodeller enbart kan göra matematiska beräkningar på siffror, måste vi konvertera (koda) denna text till numeriska värden.

Hur vi hanterar den beror på vilken typ av kategorisk data vi har:

*   **Ordinal data:** Detta är kategorier som har en tydlig och naturlig rangordning (t.ex. *Låg, Mellan, Hög*). För att hantera detta använder vi **ordinal encoding**. Vi byter helt enkelt ut texten mot siffror som följer rangordningen (t.ex. Låg=1, Mellan=2, Hög=3).

*   **Nominal data:** Detta är kategorier som *inte* har någon inbördes rangordning (t.ex. *Sverige, Norge, Danmark*). Om vi skulle ge dem siffrorna 1, 2 och 3 skulle modellen tro att Danmark (3) är "tre gånger mer värt" än Sverige (1), vilket är fel. Därför använder vi **one-hot-encoding**. Det innebär att vi skapar en helt ny kolumn för varje land, och fyller den med 1 (sant) eller 0 (falskt).

*   **Dummy-variable-encoding:** Detta är i princip samma sak som one-hot-encoding, men med en viktig matematisk genväg: vi raderar en av de nyskapade kolumnerna. Om vi har tre länder räcker det nämligen med två kolumner. Om en person har en 0:a på "Norge" och en 0:a på "Danmark", räknar datorn automatiskt ut att personen måste bo i "Sverige". Vi gör detta för att undvika överflödig data (multikollinearitet) som annars kan förvirra vissa modeller.


**g) Vad är feature engineering för något?**
* Feature engineering är den kreativa processen där man omvandlar rådata eller skapar helt nya variabler (kolumner) utifrån sin befintliga data, för att hjälpa maskininlärningsmodellen att prestera bättre och hitta mönster lättare.

Maskininlärningsmodeller är fantastiska på matematik, men de saknar mänsklig logik. Genom feature engineering använder vi vår mänskliga förståelse för problemet för att "servera" datan på ett smartare sätt till modellen. 

*Exempel:* Om vi ska bygga en modell som förutsäger huspriser och vi har en kolumn för "Totalt antal rum" och en för "Totalt antal personer i hushållet". Istället för att skicka in dessa två var för sig, kan vi räkna ut och skapa en helt ny kolumn som heter "Rum per person". Denna nya, smartare variabel (feature) är ofta mycket mer relevant för huspriset än den ursprungliga rådatan.

**h) Vad menas med principle of parsimony?**
* "Principle of parsimony" (ofta känt som Ockhams rakkniv) är en grundregel som handlar om strävan efter enkelhet. Den säger att man inte ska göra saker mer komplicerade än vad som krävs för att lösa ett problem.

Inom maskininlärning innebär detta att om du har två (eller flera) olika modeller som ger ungefär lika bra resultat och precision, ska du alltid välja den absolut enklaste modellen. 

Varför? Jo, en enkel modell med färre "rattar" och variabler är snabbare att träna, mycket lättare för oss människor att tolka, och framförallt löper den betydligt mindre risk att drabbas av överanpassning (overfitting) jämfört med en onödigt komplex modell.


4. Vad menas med att “en modell är en förenkling av verkligheten”?**
* Verkligheten är oändligt komplex och påverkas av miljontals små faktorer, slump och brus som vi omöjligt kan mäta perfekt. En maskininlärningsmodell försöker inte fånga *exakt allt*, utan den skalar bort onödigt brus för att bara fokusera på de allra viktigaste mönstren och sambanden. 

Om vi skulle försöka bygga en modell som är en exakt kopia av verkligheten, med alla dess undantag, skulle modellen bli alldeles för komplicerad och oflexibel. En modell är alltså en kompromiss – en praktisk och förenklad "karta" över verkligheten som är tillräckligt bra för att lösa vårt problem, även om den aldrig kan bli 100 % perfekt. (Det finns ett klassiskt citat inom statistiken som lyder: *"Alla modeller är fel, men vissa är användbara"*).

5. Vad menas med att en modell är “överanpassad” eller overfitted på engelska?**
*Svar:* En modell är överanpassad när den har lärt sig träningsdatan *för* bra. Istället för att bara hitta de stora, generella sambanden, har den börjat memorera varje liten detalj, inklusive slumpmässigt brus och extrema undantag (outliers) i just den specifika datan.

Resultatet av överanpassning är att modellen presterar fantastiskt bra på sin träningsdata, men misslyckas helt när den testas på ny, osedd data. 

*Liknelse:* Det är precis som en student som memorerar de exakta svaren på ett gammalt övningsprov, istället för att försöka förstå det underliggande ämnet. Studenten får alla rätt på övningsprovet (träningsdatan), men kuggar totalt när det riktiga provet (testdatan) kommer med nya, något omformulerade frågor.

6.  Högre är bättre i scikit-learn scoring, vad innebär det?**
*Svar:* I scikit-learn är alla optimeringsverktyg (som till exempel `GridSearchCV`) programmerade med en enda enkel regel: *Försök alltid att maximera poängen (få ett så högt värde som möjligt).*

Detta fungerar jättebra för mått som *Accuracy* (träffsäkerhet), där 95% är bättre än 80%. Men det skapar ett problem när vi mäter *fel* med till exempel MSE eller RMSE. Där vill vi ju att värdet ska vara så lågt som möjligt (noll fel är det allra bästa).

För att inte förvirra algoritmen och bryta mot "högre är bättre"-regeln, löser scikit-learn detta genom att göra felmåtten till **negativa tal** (t.ex. `neg_mean_squared_error`). 
Ett fel på 10 blir alltså -10. Eftersom -10 är ett matematiskt *större* (högre) tal än -50, kan algoritmen fortsätta sträva efter att få det "högsta" möjliga värdet på skalan, och systemet förblir konsekvent oavsett vilket mått vi använder.

c. Vad står RMSE för och vad mäter det?
RMSE står för Root Mean Squared Error. Det är ett utvärderingsmått för regressionsmodeller som visar hur mycket modellens förutsägelser i genomsnitt avviker från de verkliga värdena. (Man tar medelfelet i kvadrat för att straffa stora fel hårdare, och drar sedan roten ur för att få tillbaka samma enhet som ursprungsdatan).

7.  Vad är tvärsnittsdata, tidsseriedata och paneldata?

A. Tvärsnittsdata (Cross-sectional data)

Vad det är: Data som samlas in för flera olika enheter (t.ex. personer, företag, hus eller länder) vid en specifik tidpunkt.

Exempel: En mätning av vad 500 olika hus i Malmö kostar just i dag. Här tittar man på skillnaderna mellan husen, men tiden (datumet) är densamma för alla.

B. Tidsseriedata (Time-series data)

Vad det är: Data som samlas in för en enda enhet (eller ett system) upprepade gånger över flera tidsperioder (t.ex. timmar, dagar, månader eller år).

Exempel: Utomhustemperaturen eller elförbrukningen för ett visst hus mätt varje timme under ett hela året (precis som i ert energiprojekt). Här är det utvecklingen över tid som står i fokus.

C. Paneldata (Panel data / Längdsnittsdata)

Vad det är: En kombination av tvärsnittsdata och tidsseriedata. Här samlar man in data för flera olika enheter över flera tidsperioder.

Exempel: Om man följer samma 500 hus (tvärsnitt) och mäter deras energiförbrukning varje månad under 5 års tid (tidsserie). Det ger en djupare bild eftersom man kan se både skillnader mellan enheterna och hur de förändras över tid.

9. Generellt sett gäller det att högre är bättre i scikit-learn scoring, därför
används exempelvis scoring='neg_mean_squared_error'. Förklara logiken bakom detta, det vill säga att vi använder “negative” mean squared
error.


Logiken bakom att använda neg_mean_squared_error hänger ihop med hur scikit-learn är uppbyggt:

Scikit-learns standardregel ("Högre är bättre"): Alla funktioner för modellutvärdering och optimering i scikit-learn (som till exempel GridSearchCV eller cross_val_score) är programmerade för att alltid maximera poängen. De letar efter det högsta värdet för att avgöra vilken modell som är bäst.

Problemet med vanlig MSE: Mean Squared Error (MSE) mäter fel (loss). Det innebär att en perfekt modell har ett MSE på 0, och ju sämre modellen är, desto högre blir det positiva talet. Om man försökte "maximera" en vanlig MSE skulle optimeringsverktyget av misstag välja den sämsta modellen (den med störst fel).


Lösningen med det negativa värdet: Genom att sätta ett minustecken framför (neg_mean_squared_error, det vill säga multiplicera MSE med -1) vänder man på skalan:

Ett litet fel (t.ex. MSE = 2) blir -2.

Ett stort fel (t.ex. MSE = 50) blir -50.

Varför det fungerar: Matematiskt sett är -2 ett högre (större) tal än -50. Genom att göra värdet negativt kan scikit-learns funktioner fortsätta följa sin regel om att maximera poängen – eftersom det högsta negativa talet i praktiken är det som ligger närmast noll och har lägst fel.


11. I koden nedan använder vi först test_size=0.2 och sedan test_size=0.25,
förklara varför det ger oss proportionerna 60-20-20.

Testdata (20%): Den första uppdelningen tar bort 20% av den totala datan som testset, vilket lämnar kvar 80% (X_train_full).   Valideringsdata (20%): Nästa uppdelning tar 25% av denna kvarvarande data (80%): $0{,}25 \times 80\% = 20\%$ (X_val).   Träningsdata (60%): Det som blir kvar av den ursprungliga datan är då $80\% - 20\% = 60\%$ (X_train).   

Det ger proportionerna 60-20-20 (träning, validering, test).
