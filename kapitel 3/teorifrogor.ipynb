2. Förklara utvärderingsmåtten RMSE, MSE och MAE.

MAE (Mean Absolute Error): Det genomsnittliga felet. Visar exakt hur fel modellen har i snitt, utan att överdriva stora fel. Lätt att förstå!

MSE (Mean Squared Error): Kvadrerat medelfel. Eftersom felen multipliceras med sig själva straffas stora fel väldigt hårt. Bra om man vill undvika riktigt stora missar.

RMSE (Root Mean Squared Error): Kvadratroten ur MSE. Gör om MSE så att det får samma enhet som ursprungsdatan igen, men behåller egenskapen att stora fel märks mycket.

3. Om vi ska rangordna olika modeller, spelar det någon roll om RMSE eller
MSE används? Varför?

Nej, det spelar ingen roll. Eftersom RMSE är kvadratroten ur MSE ger båda måtten exakt samma inbördes ordning (ranking) mellan modellerna. Modellen med lägst MSE har alltid lägst RMSE.

4. Förklara mycket översiktligt vad gradient descent är.

Gradient descent är en optimeringsmetod som används för att lära modellen att göra så små fel som möjligt.

5. Vad är the bias variance trade-off? Varför är mer komplexa modeller inte
alltid bättre?

Bias (Systematiskt fel): Uppstår om modellen är för enkel och missar viktiga mönster i datan (underfitting).

Variance (Varians): Uppstår om modellen är för komplex och blir så känslig att den börjar lära sig slumpmässigt "brus" i träningsdatan i stället för de verkliga mönstren (overfitting).

Varför är mer komplexa modeller inte alltid bättre?
Därför att de har för hög varians. En alltför komplex modell presterar ofta perfekt på sin träningsdata (eftersom den memorerar den), men den förlorar sin förmåga att generalisera. När den ställs inför ny, okänd data i verkligheten misslyckas den därför kapitalt. Målet är att hitta en gyllene medelväg!


8. Vad är skillnaden mellan bagging och pasting?

Bagging (Bootstrap Aggregating): Tränar modellerna på slumpmässiga delmängder av träningsdatan med återläggning. Det betyder att samma dataobjekt kan väljas flera gånger till samma urval.

Pasting: Fungerar på samma sätt, men urvalet görs utan återläggning. Det betyder att varje dataobjekt bara kan väljas en enda gång per delmängd.
