2. Förklara utvärderingsmåtten RMSE, MSE och MAE.

MAE (Mean Absolute Error): Det genomsnittliga felet. Visar exakt hur fel modellen har i snitt, utan att överdriva stora fel. Lätt att förstå!

MSE (Mean Squared Error): Kvadrerat medelfel. Eftersom felen multipliceras med sig själva straffas stora fel väldigt hårt. Bra om man vill undvika riktigt stora missar.

RMSE (Root Mean Squared Error): Kvadratroten ur MSE. Gör om MSE så att det får samma enhet som ursprungsdatan igen, men behåller egenskapen att stora fel märks mycket.

3. Om vi ska rangordna olika modeller, spelar det någon roll om RMSE eller
MSE används? Varför?

Nej, det spelar ingen roll. Eftersom RMSE är kvadratroten ur MSE ger båda måtten exakt samma inbördes ordning (ranking) mellan modellerna. Modellen med lägst MSE har alltid lägst RMSE.

4. Förklara mycket översiktligt vad gradient descent är.

Gradient descent är en optimeringsmetod som används för att lära modellen att göra så små fel som möjligt.

5. Vad är the bias variance trade-off? Varför är mer komplexa modeller inte
alltid bättre?

Bias (Systematiskt fel): Uppstår om modellen är för enkel och missar viktiga mönster i datan (underfitting).

Variance (Varians): Uppstår om modellen är för komplex och blir så känslig att den börjar lära sig slumpmässigt "brus" i träningsdatan i stället för de verkliga mönstren (overfitting).

Varför är mer komplexa modeller inte alltid bättre?
Därför att de har för hög varians. En alltför komplex modell presterar ofta perfekt på sin träningsdata (eftersom den memorerar den), men den förlorar sin förmåga att generalisera. När den ställs inför ny, okänd data i verkligheten misslyckas den därför kapitalt. Målet är att hitta en gyllene medelväg!


6. Översiktlig förklaring av regressionsmodeller:

a) Linjär regression: Försöker dra en rät linje (hyperplan) genom datan som minimerar avståndet (felet) till alla datapunkter.

b) Ridge regression: En linjär regression men med L2-regularisering. Den lägger till ett straff för stora koefficientvikter, vilket tvingar modellen att hålla vikterna små och spridda, och minskar därmed risken för överanpassning.

c) Lasso regression: Linjär regression med L1-regularisering. Den straffar också stora vikter, men kan driva vissa vikter till exakt noll. Den fungerar därmed som inbyggd variabelselektion (feature selection) och rensar bort oviktiga variabler.

d) Elastic net: En blandning av Ridge och Lasso. Den kombinerar både L1- och L2-straff, vilket gör att man kan styra balansen mellan att krympa vikter (Ridge) och att eliminera variabler helt (Lasso).

e) Support vector machines (SVR): Försöker passa in så många datapunkter som möjligt inuti en "marginal" (en tub) runt regressionslinjen, till skillnad från vanlig linjär regression som försöker minimera avståndet för alla punkter. Punkter utanför tuben är de som styr felet (support vectors).

f) Beslutsträd (Regression): Delar upp datan i allt mindre rutor (regioner) baserat på if/else-regler för olika variabler. Prediktionen för en ny datapunkt blir genomsnittet av träningsdatans värden i den slutgiltiga rutan (lövet) den hamnar i.

g) Ensemble learning (Voting/Bagging): Kombinerar flera olika (eller likadana) maskininlärningsmodeller. För regression tar man vanligtvis genomsnittet av alla ingående modellers prediktioner. Det brukar ge stabilare och bättre resultat än en enskild modell.

h) Random forest (Regression): En specifik ensemble-metod (bagging) som bygger många olika beslutsträd på slumpmässiga urval av datan och på slumpmässiga urval av variabler. Slutprediktionen är snittet av alla trädens prediktioner.

7. Vad menas med white box modeller och black box modeller?

White box modeller: Modeller där det är tydligt och enkelt för en människa att förstå hur modellen kom fram till sitt beslut (exempelvis Linjär regression eller ett enkelt Beslutsträd). Formlerna och reglerna är transparenta.

Black box modeller: Komplexa modeller (som Random Forest eller Neurala Nätverk) där det är väldigt svårt eller omöjligt att i detalj spåra den exakta matematiska vägen från indata till utdata. Man matar in data och får ett svar, men ser inte tydligt vad som hände inuti "den svarta lådan".

8. Vad är skillnaden mellan bagging och pasting?

Bagging (Bootstrap Aggregating): Tränar modellerna på slumpmässiga delmängder av träningsdatan med återläggning. Det betyder att samma dataobjekt kan väljas flera gånger till samma urval.

Pasting: Fungerar på samma sätt, men urvalet görs utan återläggning. Det betyder att varje dataobjekt bara kan väljas en enda gång per delmängd.

9. Hur man kan tolka figur 3.1 på sidan 113.
(Generellt svar baserat på vanliga regressions-figurer i AI-böcker):

Figuren visar troligtvis en enkel linjär regression där en rät linje har anpassats till ett spridningsdiagram (scatter plot) av datapunkter. Linjen representerar modellens prediktioner. De små avstånden (vanligtvis inritade som vertikala streck) från de faktiska datapunkterna till den dragna linjen representerar felen (residuals). Modellens mål är att hitta lutningen och placeringen på linjen som gör summan av dessa fel så liten som möjligt (oftast genom minsta kvadratmetoden, MSE).
