2. Vad menas med att en modell produktionssätts?**

Att produktionssätta (ofta kallat "deployment") innebär att man tar den färdigtränade modellen från sin isolerade utvecklingsmiljö (som en Jupyter Notebook) och integrerar den i en verklig, fungerande mjukvara. 

Istället för att bara utvärderas på sparad testdata, kopplas modellen nu upp för att ta emot helt ny, levande data från riktiga användare och leverera prediktioner i realtid. Det kan till exempel vara att bygga in modellen i en banksystem-backend för att i realtid blockera bedrägliga kortköp, eller att integrera den i en webbapp (typ Hemnet) för att automatiskt värdera hus så fort en användare knappar in sina kvadratmeter.

3. Vad är scikit-learn för något? Biblioteket följer några centrala designprinciper. Vilka är dessa? Vad är estimators, predictors och transformers?**

* **Scikit-learn:** Det är ett av de mest populära open-source biblioteken för maskininlärning i Python. Det innehåller massor av färdigskrivna algoritmer för regression, klassificering och dataförberedelse så att vi slipper bygga all komplicerad matte från noll.
* **Designprinciper:** Huvudprincipen är *enhetlighet*. Scikit-learn har ett extremt konsekvent gränssnitt. Har du lärt dig hur en modell fungerar, vet du i princip hur alla andra fungerar också. Biblioteket bygger på tre tydliga byggstenar (objekt):
* **Estimators (Beräknare):** Vilket objekt som helst som kan *lära sig* mönster eller parametrar från data. Alla estimators använder metoden `.fit(data)` för att utföra sin inlärning.
* **Transformers (Omvandlare):** Vissa estimators är "transformers". Deras jobb är att ta rådata och modifiera (tvätta) den, till exempel genom att skala om siffror eller översätta text till siffror (One-Hot-Encoding). De använder metoden `.transform(data)`.
* **Predictors (Förutsägare):** Andra estimators är "predictors", alltså själva AI-modellerna (som Linjär Regression eller Beslutsträd). När de har lärt sig datan via `fit()`, kan de göra gissningar på helt ny data med hjälp av metoden `.predict(ny_data)`.

4. Vad är TensorFlow och Keras?**
 
* **TensorFlow:** Det är ett stort och extremt kraftfullt open-source ramverk för maskininlärning, skapat av Google. Det används i första hand för *Deep Learning* (djupinlärning) och artificiella neurala nätverk. TensorFlow är själva "motorn" som hanterar alla de tunga, komplexa matematiska beräkningarna i bakgrunden (ofta via grafikkort, GPUs).
* **Keras:** Det är ett så kallat högnivå-API som körs *ovanpå* TensorFlow. Eftersom TensorFlows egen grundkod kan vara ganska krånglig och svårläst, skapades Keras för att vara mycket mer mänskligt och användarvänligt. Om TensorFlow är motorn i bilen, är Keras ratten och pedalerna som gör det enkelt och intuitivt för oss utvecklare att bygga neurala nätverk. (Idag är Keras standardgränssnittet och är helt inbyggt i TensorFlow).

8. Förklara vad koden nedan gör. Varför är det viktigt att kunna spara en modell?**

*Svar:* 
Här är en rad-för-rad förklaring av vad koden gör:
1. `from ... import ...` (Rad 1-3): Importerar de verktyg som behövs, inklusive `dump` och `load` från biblioteket `joblib` som används för att spara filer.
2. `X, y = make_regression(...)`: Skapar ett "låtsas-dataset" med 20 000 rader och 3 kolumner för att ha något att öva på.
3. `model = LinearRegression().fit(X, y)`: Skapar en linjär regressionsmodell och tränar den direkt på datan.
4. `dump(model, "linear_model.joblib")`: **Sparar** den färdigtränade modellen på din hårddisk i en fil som heter "linear_model.joblib".
5. `model_loaded = load(...)`: **Laddar in** modellen från filen igen till en ny variabel.
6. `print(model_loaded.predict(X[:5]))`: Använder den inlästa modellen för att göra en prediktion på de fem första raderna i datan.

**Varför är det viktigt att kunna spara en modell?**
Att träna en avancerad maskininlärningsmodell på miljontals rader data kan ta timmar, dagar eller till och med veckor, och det kräver enormt mycket datorkraft. Om vi inte kunde spara den inlärda modellen, skulle vi behöva göra om all denna tunga träning varje gång programmet startades om eller varje gång vi ville använda modellen i en app (produktionssättning). Genom att spara den som en fil (t.ex. `.joblib` eller `.pkl`) kapslar vi in all kunskap den lärt sig, så att den är redo att användas blixtsnabbt när som helst.

```python
from sklearn.datasets import make_regression
from sklearn.linear_model import LinearRegression
from joblib import dump, load

X, y = make_regression(n_samples=20000, n_features=3, noise=0.1)
model = LinearRegression().fit(X, y)
dump(model, "linear_model.joblib")

model_loaded = load("linear_model.joblib")
print("Prediktioner från den inlästa filen:")
print(model_loaded.predict(X[:5]))
```
