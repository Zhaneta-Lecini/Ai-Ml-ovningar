1. Vad är klustring för något? Ge några exempel på tillämpningsområden.

Klustring är en form av oövervakad maskininlärning (unsupervised learning) där algoritmen grupperar datapunkter som liknar varandra utan att veta vad grupperna i förväg representerar.
Exempel på användning: Kundsegmentering (dela in kunder baserat på köpbeteende), anomalidetektion (hitta avvikande banktransaktioner/bedrägerier) och rekommendationssystem.

Klustring skiljer sig från allt vi gjort hittills eftersom vi inte har något "rätt svar" (ingen y-variabel) att träna på. Vi ber bara algoritmen hitta mönster och gruppera data på egen hand (kallas oövervakad inlärning).

2. Förklara översiktligt hur K-means fungerar.

K-means placerar slumpmässigt ut ett bestämt antal (k) mittpunkter, kallade centroider, i datan. Varje datapunkt tilldelas sedan den centroid den ligger närmast, vilket bildar initiala kluster. Därefter beräknar modellen den exakta mittpunkten för varje nytt kluster och flyttar centroiden dit. Denna process (tilldela punkter -> flytta centroider) upprepas tills centroiderna stannar och klustren stabiliserats.

3. Hur kan man välja vilket antal kluster som ska användas för en K-means modell?
Man kan använda flera metoder för att hitta det optimala antalet kluster:

Inertia: Mäter hur långt datapunkterna har till sin centroid (lägre är bättre). Man plottar detta och letar efter en "armbåge" (Elbow method) där minskningen plötsligt planar ut.

Silhouette score: Ett mått mellan -1 och 1 som visar hur väl en punkt passar i sitt eget kluster jämfört med angränsande kluster. Ju närmare 1, desto bättre och tydligare separerade är klustren.

Silhouette diagram: En grafisk representation av silhouette scores för alla kluster. Man vill att alla kluster ("knivbladen") ska passera det genomsnittliga värdet och vara någorlunda lika tjocka (innehålla lika många datapunkter).
